```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns



sns.set_style("whitegrid")


print("Welcome to Netflix Data Analysis and Visualization Project!")
print("All libraries imported successfully!")


import zipfile

with zipfile.ZipFile('/content/netflix_titles.csv.zip', 'r') as zip_ref:
    zip_ref.extractall('/content/')

print("Dataset extracted successfully!")
df = pd.read_csv('/content/netflix_titles.csv')

df.head()
df = pd.read_csv('/content/netflix_titles.csv')



df.head()



print("Number of Rows and Columns:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()

# Statistical Summary

df.describe(include='all')
# Checking Missing Values

print("Missing Values in Each Column:")

print(df.isnull().sum())
# Checking Duplicate Records

print("Total Duplicate Records:", df.duplicated().sum())

# Removing Duplicate Records

df.drop_duplicates(inplace=True)

print("Duplicates Removed Successfully!")

print("Updated Dataset Shape:", df.shape)

# Checking Missing Values

print("Missing Values Before Cleaning:")
print(df.isnull().sum())

# Filling Missing Values

df['director'] = df['director'].fillna('Unknown')

df['cast'] = df['cast'].fillna('Unknown')

df['country'] = df['country'].fillna('Unknown')

df['rating'] = df['rating'].fillna(df['rating'].mode()[0])

df['date_added'] = df['date_added'].fillna('Not Available')

df['duration'] = df['duration'].fillna('Not Available')

print("\nMissing Values Handled Successfully!")

# Final Dataset Check

print("Final Dataset Shape:", df.shape)

print("\nRemaining Missing Values:")
print(df.isnull().sum())

print("\nCleaned Dataset Preview:")
df.head()
# Movies vs TV Shows Analysis

content_counts = df['type'].value_counts()

plt.figure(figsize=(8,5))

sns.barplot(
    x=content_counts.index,
    y=content_counts.values
)

plt.title('Distribution of Movies and TV Shows on Netflix',
          fontsize=14)

plt.xlabel('Content Type')
plt.ylabel('Number of Titles')

plt.show()

# Movies vs TV Shows Pie Chart

plt.figure(figsize=(7,7))

plt.pie(
    content_counts.values,
    labels=content_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Movies vs TV Shows Percentage Distribution')

plt.show()

print("Netflix Content Distribution:")
print(content_counts)

print("\nPercentage Distribution:")
print((content_counts / content_counts.sum() * 100).round(2))

# Top 10 Countries with Most Netflix Content

country_counts = df['country'].str.split(', ').explode()

top_countries = country_counts[
    country_counts != 'Unknown'
].value_counts().head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_countries.values,
    y=top_countries.index
)

plt.title('Top 10 Countries with Most Netflix Content', fontsize=16)

plt.xlabel('Number of Movies and TV Shows')
plt.ylabel('Country')

plt.show()

print("Top 10 Countries on Netflix:")

print(top_countries)
# Netflix Content Growth Over the Years

year_counts = df['release_year'].value_counts().sort_index()

plt.figure(figsize=(14,6))

plt.plot(
    year_counts.index,
    year_counts.values,
    marker='o'
)

plt.title('Netflix Content Growth Over the Years', fontsize=16)

plt.xlabel('Release Year')
plt.ylabel('Number of Movies and TV Shows')

plt.grid(True)

plt.show()

# Top 10 Years with Most Content

top_years = df['release_year'].value_counts().head(10)

print("Top 10 Release Years:")
print(top_years)

# Analyzing Top 10 Netflix Genres

genre_counts = df['listed_in'].dropna().str.split(', ').explode()

top_genres = genre_counts.value_counts().head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_genres.values,
    y=top_genres.index
)

plt.title('Top 10 Popular Genres on Netflix', fontsize=16)

plt.xlabel('Number of Titles')
plt.ylabel('Genres')

plt.show()

print("Top 10 Genres on Netflix:")

print(top_genres)

rating_counts = df['rating'].value_counts().head(10)

plt.figure(figsize=(12,6))
sns.barplot(x=rating_counts.values, y=rating_counts.index)

plt.title('Top 10 Content Ratings on Netflix', fontsize=16)
plt.xlabel('Number of Titles')
plt.ylabel('Content Rating')
plt.show()

print("Top 10 Netflix Content Ratings:")
print(rating_counts)


release_year_counts = df['release_year'].value_counts().sort_index()

plt.figure(figsize=(14,6))
plt.plot(release_year_counts.index, release_year_counts.values)

plt.title('Netflix Content Release Year Analysis', fontsize=16)
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(True)
plt.show()

print("Top 10 Release Years on Netflix:")
print(df['release_year'].value_counts().head(10))

movies = df[df['type'] == 'Movie'].copy()

movies['duration_minutes'] = movies['duration'].str.extract('(\d+)').astype(float)

plt.figure(figsize=(12,6))
sns.histplot(movies['duration_minutes'].dropna(), bins=30, kde=True)

plt.title('Distribution of Movie Durations on Netflix', fontsize=16)
plt.xlabel('Duration (Minutes)')
plt.ylabel('Number of Movies')
plt.show()

print("Average Movie Duration:")
print(round(movies['duration_minutes'].mean(), 2), "minutes")

print("\nShortest Movie:")
print(movies['duration_minutes'].min(), "minutes")

print("\nLongest Movie:")
print(movies['duration_minutes'].max(), "minutes")

tv_shows = df[df['type'] == 'TV Show'].copy()

tv_shows['seasons'] = tv_shows['duration'].str.extract(r'(\d+)').astype(float)

print("Total TV Shows:", len(tv_shows))

season_counts = tv_shows['seasons'].value_counts().sort_index()

plt.figure(figsize=(12,6))
sns.barplot(x=season_counts.index, y=season_counts.values)

plt.title('TV Shows Seasons Distribution on Netflix', fontsize=16)
plt.xlabel('Number of Seasons')
plt.ylabel('Number of TV Shows')
plt.show()

print("TV Shows by Number of Seasons:")
print(season_counts.sort_values(ascending=False).head(10))

country_counts = df['country'].dropna().str.split(', ').explode().value_counts().head(10)

plt.figure(figsize=(12,6))
sns.barplot(x=country_counts.values, y=country_counts.index)

plt.title('Top 10 Countries by Netflix Content', fontsize=16)
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.show()
print("Top 10 Countries with Most Netflix Content:")
print(country_counts)


country_type = df.dropna(subset=['country']).copy()

country_type['country'] = country_type['country'].str.split(', ')

country_type = country_type.explode('country')

country_comparison = pd.crosstab(
    country_type['country'],
    country_type['type']
)

top_countries = country_type['country'].value_counts().head(10).index

country_comparison = country_comparison.loc[top_countries]

print(country_comparison)

country_comparison.plot(
    kind='bar',
    figsize=(14,7)
)

plt.title('Movies vs TV Shows by Top 10 Countries', fontsize=16)
plt.xlabel('Country')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.legend(title='Content Type')
plt.tight_layout()
plt.show()

df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), errors='coerce')

df['year_added'] = df['date_added'].dt.year

yearly_content = df.groupby(['year_added', 'type']).size().unstack(fill_value=0)

print(yearly_content.tail(15))

yearly_content.plot(
    figsize=(14, 7),
    marker='o'
)

plt.title('Netflix Content Addition Trend Over the Years', fontsize=16)
plt.xlabel('Year')
plt.ylabel('Number of Titles Added')
plt.legend(title='Content Type')
plt.grid(True)
plt.tight_layout()
plt.show()

print("Year with Most Content Added:")

print(yearly_content.sum(axis=1).idxmax())

print("\nTotal Content Added in That Year:")

print(yearly_content.sum(axis=1).max())

genre_data = df.dropna(subset=['listed_in']).copy()

genre_data['listed_in'] = genre_data['listed_in'].str.split(', ')

genre_data = genre_data.explode('listed_in')

genre_counts = genre_data['listed_in'].value_counts().head(10)

print("Top 10 Netflix Genres:")
print(genre_counts)

plt.figure(figsize=(12,6))

sns.barplot(
    x=genre_counts.values,
    y=genre_counts.index
)

plt.title('Top 10 Most Popular Genres on Netflix', fontsize=16)
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
plt.tight_layout()
plt.show()

genre_comparison = pd.crosstab(
    genre_data['listed_in'],
    genre_data['type']
)

top_10_genres = genre_counts.index

genre_comparison = genre_comparison.loc[top_10_genres]

print("Movies vs TV Shows Genre Comparison:")
print(genre_comparison)

genre_comparison.plot(
    kind='bar',
    figsize=(14,7)
)

plt.title('Movies vs TV Shows by Genre on Netflix', fontsize=16)
plt.xlabel('Genre')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Content Type')
plt.tight_layout()
plt.show()

rating_data = df.dropna(subset=['rating']).copy()

rating_comparison = pd.crosstab(
    rating_data['rating'],
    rating_data['type']
)

top_ratings = rating_comparison.sum(axis=1).nlargest(10).index

rating_comparison = rating_comparison.loc[top_ratings]

print("Content Rating Comparison:")
print(rating_comparison)

rating_comparison.plot(
    kind='bar',
    figsize=(14, 7)
)

plt.title('Content Ratings: Movies vs TV Shows', fontsize=16)
plt.xlabel('Content Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.legend(title='Content Type')
plt.tight_layout()
plt.show()

director_data = df.dropna(subset=['director']).copy()

director_data['director'] = director_data['director'].str.split(',')

director_data = director_data.explode('director')

director_data['director'] = director_data['director'].str.strip()

director_counts = director_data['director'].value_counts().head(10)

print("Top 10 Directors on Netflix:")
print(director_counts)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=director_counts.values,
    y=director_counts.index
)

plt.title('Top 10 Directors on Netflix', fontsize=16)
plt.xlabel('Number of Titles')
plt.ylabel('Director')
plt.tight_layout()
plt.show()

actor_data = df.dropna(subset=['cast']).copy()

actor_data['cast'] = actor_data['cast'].str.split(',')

actor_data = actor_data.explode('cast')

actor_data['cast'] = actor_data['cast'].str.strip()

actor_counts = actor_data['cast'].value_counts().head(10)

print("Top 10 Actors on Netflix:")
print(actor_counts)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=actor_counts.values,
    y=actor_counts.index
)

plt.title('Top 10 Actors on Netflix', fontsize=16)
plt.xlabel('Number of Titles')
plt.ylabel('Actor')
plt.tight_layout()
plt.show()

# Total Movies and TV Shows
total_titles = len(df)

movies_count = (df['type'] == 'Movie').sum()
tv_count = (df['type'] == 'TV Show').sum()

# Top Content Rating
top_rating_name = df['rating'].dropna().value_counts().idxmax()

# Top Country
country_names = (
    df['country']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
)

top_country_name = country_names.value_counts().idxmax()

# Top Genre
genre_names = (
    df['listed_in']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
)

top_genre_name = genre_names.value_counts().idxmax()

# Average Movie Duration
movie_durations = pd.to_numeric(
    df.loc[df['type'] == 'Movie', 'duration']
    .astype(str)
    .str.extract(r'(\d+)')[0],
    errors='coerce'
)

average_duration = round(movie_durations.mean(), 2)

# Most Content Added Year
added_dates = pd.to_datetime(
    df['date_added'].astype('string').str.strip(),
    errors='coerce'
)

added_years = added_dates.dt.year.value_counts()

most_added_year = added_years.idxmax()

# Display Insights
print("===== NETFLIX DATA ANALYSIS: KEY INSIGHTS =====")

print("Total Titles:", total_titles)
print("Total Movies:", movies_count)
print("Total TV Shows:", tv_count)

print("Most Common Content Rating:", top_rating_name)
print("Top Country:", top_country_name)
print("Top Genre:", top_genre_name)

print("Average Movie Duration:", average_duration, "minutes")

print("Year with Most Content Added:", most_added_year)

# Prepare Dashboard Data

dashboard_types = df['type'].value_counts()

dashboard_countries = (
    df['country']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)

dashboard_genres = (
    df['listed_in']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)

dashboard_ratings = df['rating'].value_counts().head(10)

# Create Dashboard

fig, axes = plt.subplots(2, 2, figsize=(18, 12))

# Chart 1: Movies vs TV Shows
dashboard_types.plot(
    kind='bar',
    ax=axes[0, 0]
)

axes[0, 0].set_title('Movies vs TV Shows')
axes[0, 0].set_xlabel('Content Type')
axes[0, 0].set_ylabel('Number of Titles')
axes[0, 0].tick_params(axis='x', rotation=0)

# Chart 2: Top Countries
axes[0, 1].barh(
    dashboard_countries.index[::-1],
    dashboard_countries.values[::-1]
)

axes[0, 1].set_title('Top 10 Countries')
axes[0, 1].set_xlabel('Number of Titles')

# Chart 3: Top Genres
axes[1, 0].barh(
    dashboard_genres.index[::-1],
    dashboard_genres.values[::-1]
)

axes[1, 0].set_title('Top 10 Genres')
axes[1, 0].set_xlabel('Number of Titles')

# Chart 4: Top Ratings
dashboard_ratings.plot(
    kind='bar',
    ax=axes[1, 1]
)

axes[1, 1].set_title('Top 10 Content Ratings')
axes[1, 1].set_xlabel('Rating')
axes[1, 1].set_ylabel('Number of Titles')
axes[1, 1].tick_params(axis='x', rotation=45)

# Dashboard Title
plt.suptitle(
    'NETFLIX DATA ANALYSIS DASHBOARD',
    fontsize=20
)

plt.tight_layout()

# Save Dashboard
plt.savefig(
    'netflix_dashboard.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()

print("""
========================================
       NETFLIX DATA ANALYSIS
          PROJECT CONCLUSION
========================================
""")

print("1. Total Netflix Titles:", total_titles)

print("2. Total Movies:", movies_count)

print("3. Total TV Shows:", tv_count)

print("4. Most Common Content Rating:", top_rating_name)

print("5. Country with Most Content:", top_country_name)

print("6. Most Common Genre:", top_genre_name)

print("7. Average Movie Duration:", average_duration, "minutes")

print("8. Highest Content Addition Year:", most_added_year)

print("""
========================================
          PROJECT COMPLETED
========================================
""")
```
